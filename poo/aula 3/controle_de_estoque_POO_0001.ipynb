```python
class Produto:
    # O método __init__ é chamado assim que criamos um novo produto
    def __init__(self, nome, estoque_inicial, vendas):
        self.nome = nome
        self.estoque_inicial = estoque_inicial
        self.vendas = vendas

    # Um método simples para calcular o estoque final
    def calcular_estoque_final(self):
        return self.estoque_inicial - self.vendas

    # Um método simples para verificar a situação do estoque
    def verificar_situacao(self):
        # Chamamos o método acima usando o 'self' para pegar o valor atualizado
        estoque_final = self.calcular_estoque_final()

        if estoque_final >= 10:
            return 'Estoque suficiente'
        elif estoque_final >= 1:
            return 'Estoque baixo'
        else:
            return 'Produto esgotado'

    # Método para imprimir os dados deste produto específico
    def exibir_relatorio(self):
        print('\n RELATÓRIO DE ESTOQUE')
        print(f'Produto: {self.nome}')
        print(f'Estoque inicial: {self.estoque_inicial}')
        print(f'Unidades vendidas: {self.vendas}')
        print(f'Estoque final: {self.calcular_estoque_final()}')
        print(f'Situação: {self.verificar_situacao()}')


# ==========================================
# FLUXO PRINCIPAL DO PROGRAMA (FORA DA CLASSE)
# ==========================================

produtos = []

while True:
    print("\n--- CADASTRO DE PRODUTO ---")

    # Coletamos os dados de forma procedural, como os alunos já sabem
    nome_digitado = input('Digite o nome do produto: ')
    inicial_digitado = int(input('Digite a quantidade inicial em estoque: '))
    vendas_digitadas = int(input('Digite quantas unidades foram vendidas: '))

    # MÁGICA DA POO: Criamos um objeto "Produto" entregando as variáveis digitadas
    novo_produto = Produto(nome_digitado, inicial_digitado, vendas_digitadas)

    # Guardamos o objeto inteiro na lista
    produtos.append(novo_produto)

    continuar = input("\nDeseja cadastrar outro produto? (s/n): ").strip().lower()
    if continuar != "s":
        break

print("\n============= RELATÓRIO GERAL =============")

# Percorremos a lista de objetos e pedimos para cada um exibir seu próprio relatório
for p in produtos:
    p.exibir_relatorio()
```

```python
# EXPLICAÇÃO: A palavra 'class' serve para criar um "molde" ou "planta" de um novo tipo de dado.
# INTENÇÃO: Agrupar tudo o que pertence a um produto (dados e ações) em um único lugar.
# DIFERENÇA PROCEDURAL: No procedural, os dados ficavam soltos em dicionários e as ações em funções isoladas. Na POO, unimos os dois.
class Produto:

    # EXPLICAÇÃO: O método mágico '__init__' é o construtor. Ele é executado automaticamente quando criamos um novo produto. O 'self' é obrigatório e representa o próprio objeto que está sendo criado.
    # INTENÇÃO: Definir quais informações o produto precisa ter logo no momento em que "nasce".
    # DIFERENÇA PROCEDURAL: Antes, capturávamos variáveis e passávamos para várias funções. Agora, guardamos dentro do próprio objeto usando o 'self'.
    def __init__(self, nome, estoque_inicial, vendas):

        # EXPLICAÇÃO: 'self.nome' cria uma variável interna (atributo) do objeto e guarda o valor recebido.
        self.nome = nome

        # EXPLICAÇÃO: Guarda a quantidade inicial dentro do objeto.
        self.estoque_inicial = estoque_inicial

        # EXPLICAÇÃO: Guarda as vendas dentro do objeto.
        self.vendas = vendas

    # EXPLICAÇÃO: Função dentro de uma classe é chamada de "método". O 'self' permite que ela acesse os dados do objeto.
    # INTENÇÃO: Realizar o cálculo do estoque usando os próprios dados do produto.
    # DIFERENÇA PROCEDURAL: Antes, a função precisava receber parâmetros: 'def calcular_estoque(inicial, vendas)'. Na POO, o método não pede parâmetros, ele pega do 'self': 'self.estoque_inicial'.
    def calcular_estoque_final(self):
        return self.estoque_inicial - self.vendas

    # EXPLICAÇÃO: Outro método do objeto. Novamente, usa 'self' para ter acesso a tudo daquela classe.
    # INTENÇÃO: Avaliar a situação do estoque com base no cálculo do estoque final.
    def verificar_situacao(self):

        # EXPLICAÇÃO: Chamamos o método 'calcular_estoque_final' usando o 'self' para descobrir o valor atualizado.
        # INTENÇÃO: Reutilizar código. Um método pode chamar outro dentro da mesma classe!
        estoque_final = self.calcular_estoque_final()

        # EXPLICAÇÃO: Estrutura condicional normal para decidir qual texto retornar.
        if estoque_final >= 10:
            return 'Estoque suficiente'
        elif estoque_final >= 1:
            return 'Estoque baixo'
        else:
            return 'Produto esgotado'

    # EXPLICAÇÃO: Método para mostrar as informações na tela.
    # INTENÇÃO: O próprio produto sabe como se imprimir, formatando seus dados.
    # DIFERENÇA PROCEDURAL: Antes, enviávamos 5 variáveis para a função imprimir. Agora, não enviamos nada: chamamos 'produto.exibir_relatorio()' e ele mesmo busca seus dados no 'self'.
    def exibir_relatorio(self):
        print('\n RELATÓRIO DE ESTOQUE')
        print(f'Produto: {self.nome}')
        print(f'Estoque inicial: {self.estoque_inicial}')
        print(f'Unidades vendidas: {self.vendas}')
        # Chamando métodos internos diretamente no print
        print(f'Estoque final: {self.calcular_estoque_final()}')
        print(f'Situação: {self.verificar_situacao()}')


# ==========================================
# FLUXO PRINCIPAL DO PROGRAMA (FORA DA CLASSE)
# ==========================================

# EXPLICAÇÃO: Lista vazia para guardar os objetos criados.
produtos = []

# EXPLICAÇÃO: Laço de repetição infinito para o menu de cadastro.
while True:
    print("\n--- CADASTRO DE PRODUTO ---")

    # EXPLICAÇÃO: Coleta de dados clássica usando input (igual ao procedural).
    nome_digitado = input('Digite o nome do produto: ')
    inicial_digitado = int(input('Digite a quantidade inicial em estoque: '))
    vendas_digitadas = int(input('Digite quantas unidades foram vendidas: '))

    # EXPLICAÇÃO E INTENÇÃO: Esta é a grande mágica da POO! Chamamos a classe 'Produto' passando os dados. Isso cria uma "Instância" (um objeto) e roda o '__init__'.
    # DIFERENÇA PROCEDURAL: Em vez de criar um dicionário cheio de chaves, criamos um objeto inteligente que já vem com as funções de calcular embutidas nele.
    novo_produto = Produto(nome_digitado, inicial_digitado, vendas_digitadas)

    # EXPLICAÇÃO: Adiciona o objeto inteiro na lista.
    produtos.append(novo_produto)

    # EXPLICAÇÃO: Pergunta se deseja continuar e quebra o laço (break) se não quiser.
    continuar = input("\nDeseja cadastrar outro produto? (s/n): ").strip().lower()
    if continuar != "s":
        break

print("\n============= RELATÓRIO GERAL =============")

# EXPLICAÇÃO: O laço 'for' percorre a lista. Cada item 'p' agora é um OBJETO da classe Produto.
# DIFERENÇA PROCEDURAL: Em vez de tirar dados do dicionário e mandar para uma função, nós simplesmente mandamos o próprio objeto executar a sua ação 'exibir_relatorio()'.
for p in produtos:
    p.exibir_relatorio()
```
