```python
def pedir_nome_produto():
    nome = input('Digite o nome do produto: ')
    return nome

def pedir_estoque_inicial():
    estoque = int(input('Digite a quantidade inicial em estoque: '))
    return estoque

def pedir_vendas():
    vendas = int(input('Digite quantas unidades foram vendidas: '))
    return vendas

def calcular_estoque_final(inicial, vendas):
    final = inicial - vendas
    return final

def verificar_situacao(estoque_final):
    if estoque_final >= 10:
        return 'Estoque suficiente'
    elif estoque_final >= 1:
        return 'Estoque baixo'
    else:
        return 'Produto esgotado'

def exibir_relatorio(nome, inicial, vendas, final, situacao):
    print('\n RELATÓRIO DE ESTOQUE')
    print(f'Produto: {nome}')
    print(f'Estoque inicial: {inicial}')
    print(f'Unidades vendidas: {vendas}')
    print(f'Estoque final: {final}')
    print(f'Situação: {situacao}')


produtos = []

while True:
    print("\n--- CADASTRO DE PRODUTO ---")

    nome = pedir_nome_produto()
    inicial = pedir_estoque_inicial()
    vendas = pedir_vendas()
    final = calcular_estoque_final(inicial, vendas)
    situacao = verificar_situacao(final)

    produto = {
        "nome": nome,
        "estoque_inicial": inicial,
        "vendas": vendas,
        "estoque_final": final,
        "situacao": situacao
    }

    produtos.append(produto)

    continuar = input("\nDeseja cadastrar outro produto? (s/n): ").strip().lower()
    if continuar != "s":
        break

print("\n============= RELATÓRIO GERAL =============")

for p in produtos:
    exibir_relatorio(
        p["nome"],
        p["estoque_inicial"],
        p["vendas"],
        p["estoque_final"],
        p["situacao"]
    )
```

```python

def pedir_nome_produto():
    # input() mostra uma mensagem ao usuário e retorna o que ele digitou (sempre como texto/str)
    nome = input('Digite o nome do produto: ')
    # return devolve o valor para quem chamou a função; aqui devolvemos o nome digitado
    return nome


# ---------- Função: pede o estoque inicial ----------
def pedir_estoque_inicial():
    # input() retorna texto; int() converte esse texto para número inteiro
    # se o usuário digitar algo que não seja número, isso causará um erro (ValueError)
    estoque = int(input('Digite a quantidade inicial em estoque: '))
    # retorna o número inteiro lido
    return estoque


# ---------- Função: pede a quantidade vendida ----------
def pedir_vendas():
    # novamente usamos int() para garantir que 'vendas' seja um número inteiro
    vendas = int(input('Digite quantas unidades foram vendidas: '))
    return vendas


# ---------- Função: calcula o estoque final ----------
def calcular_estoque_final(inicial, vendas):
    # recebe dois parâmetros: 'inicial' e 'vendas'
    # subtrai vendas do estoque inicial para obter o estoque final
    final = inicial - vendas
    # devolve o resultado da subtração
    return final


# ---------- Função: verifica a situação do estoque ----------
def verificar_situacao(estoque_final):
    # estrutura condicional que analisa o valor numérico de 'estoque_final'
    # se estoque_final for maior ou igual a 10 -> "Estoque suficiente"
    if estoque_final >= 10:
        return 'Estoque suficiente'
    # se não, mas estoque_final for maior ou igual a 1 -> "Estoque baixo"
    elif estoque_final >= 1:
        return 'Estoque baixo'
    # caso contrário (estoque_final igual a 0 ou negativo) -> "Produto esgotado"
    else:
        return 'Produto esgotado'


# ---------- Função: exibe o relatório de um produto ----------
def exibir_relatorio(nome, inicial, vendas, final, situacao):
    # print() imprime texto no console. Aqui mostramos um cabeçalho simples para o relatório.
    print('\n=== RELATÓRIO DE ESTOQUE ===')
    # f-strings (f'...') permitem inserir valores de variáveis diretamente no texto
    print(f'Produto: {nome}')
    print(f'Estoque inicial: {inicial}')
    print(f'Unidades vendidas: {vendas}')
    print(f'Estoque final: {final}')
    print(f'Situação: {situacao}')


# =====================================================
# PROGRAMA PRINCIPAL
# =====================================================

# criamos uma lista vazia chamada 'produtos' que irá guardar os dicionários de cada produto
produtos = []

# laço while True: repete indefinidamente até encontrarmos um break
# usamos esse laço para permitir cadastrar vários produtos seguidos
while True:
    # mensagem para o usuário indicando que um novo cadastro vai começar
    print("\n--- CADASTRO DE PRODUTO ---")

    # chamamos as funções que leem os dados do usuário
    # cada função retorna um valor que guardamos em variáveis locais
    nome = pedir_nome_produto()          # string com o nome do produto
    inicial = pedir_estoque_inicial()   # inteiro com a quantidade inicial
    vendas = pedir_vendas()              # inteiro com a quantidade vendida

    # calculamos o estoque final utilizando a função dedicada
    final = calcular_estoque_final(inicial, vendas)

    # verificamos a situação (mensagem) a partir do estoque final
    situacao = verificar_situacao(final)

    # montamos um dicionário com todos os dados do produto
    # dicionário: estrutura que guarda pares chave:valor (útil para agrupar informações)
    produto = {
        "nome": nome,
        "estoque_inicial": inicial,
        "vendas": vendas,
        "estoque_final": final,
        "situacao": situacao
    }

    # adicionamos o dicionário à lista 'produtos'
    # append() coloca o item no final da lista
    produtos.append(produto)

    # perguntamos ao usuário se deseja cadastrar outro produto
    # .strip() remove espaços em branco antes/depois; .lower() converte para minúsculas
    continuar = input("\nDeseja cadastrar outro produto? (s/n): ").strip().lower()
    # se o usuário não digitar 's' (sim), saímos do loop com break
    if continuar != "s":
        break

# após sair do loop, exibimos um relatório geral com todos os produtos cadastrados
print("\n============= RELATÓRIO GERAL =============")

# percorremos a lista 'produtos' com um laço for
# cada item 'p' é um dicionário representando um produto
for p in produtos:
    # chamamos a função que exibe o relatório, passando os valores do dicionário
    exibir_relatorio(
        p["nome"],
        p["estoque_inicial"],
        p["vendas"],
        p["estoque_final"],
        p["situacao"]
    )
```
