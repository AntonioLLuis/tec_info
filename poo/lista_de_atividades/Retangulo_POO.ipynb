```python
class Terreno:
    def __init__(self, largura, comprimento, valor_metro_quadrado):
        self.largura = largura
        self.comprimento = comprimento
        self.valor_metro_quadrado = valor_metro_quadrado

    def calcular_area(self):
        return self.largura * self.comprimento

    def calcular_preco(self):
        area = self.calcular_area()
        return area * self.valor_metro_quadrado

    def mostrar_resultados(self):
        print(f"Area do terreno = {self.calcular_area():.2f}")
        print(f"Preco do terreno = {self.calcular_preco():.2f}\n")


# ==========================================
# FLUXO PRINCIPAL DO PROGRAMA
# ==========================================

largura_digitada = float(input("Digite a largura do terreno: "))
comprimento_digitado = float(input("Digite o comprimento do terreno: "))
valor_digitado = float(input("Digite o valor do metro quadrado: "))

meu_terreno = Terreno(largura_digitada, comprimento_digitado, valor_digitado)

meu_terreno.mostrar_resultados()
```

```python
# EXPLICAÇÃO: A classe funciona como uma "planta baixa" para qualquer terreno que formos criar no sistema.
# INTENÇÃO: Agrupar as dimensões (dados) e os cálculos (ações) no mesmo lugar.
# DIFERENÇA PROCEDURAL: Em vez de termos variáveis soltas pelo código, agora encapsulamos tudo dentro do conceito "Terreno".
class Terreno:

    # EXPLICAÇÃO: O método construtor '__init__' recebe os valores logo que o terreno é criado e os salva usando 'self'.
    # INTENÇÃO: Garantir que nenhum terreno nasça sem largura, comprimento e valor do metro quadrado.
    def __init__(self, largura, comprimento, valor_metro_quadrado):
        self.largura = largura
        self.comprimento = comprimento
        self.valor_metro_quadrado = valor_metro_quadrado

    # EXPLICAÇÃO: Método interno para calcular a área.
    # INTENÇÃO: O próprio terreno sabe calcular o seu tamanho.
    # DIFERENÇA PROCEDURAL: Antes fazíamos a conta solta "area = largura * comprimento". Aqui, o método não precisa de parâmetros externos, ele busca as medidas guardadas dentro dele mesmo (self).
    def calcular_area(self):
        return self.largura * self.comprimento

    # EXPLICAÇÃO: Método para calcular o preço total.
    # INTENÇÃO: Utilizar a área já calculada para definir o preço final.
    # DIFERENÇA PROCEDURAL: Veja como é elegante: usamos 'self.calcular_area()' para reaproveitar a lógica do método anterior, em vez de refazer a conta.
    def calcular_preco(self):
        area = self.calcular_area()
        return area * self.valor_metro_quadrado

    # EXPLICAÇÃO: Método que imprime os resultados formatados na tela.
    # INTENÇÃO: Deixar a responsabilidade de "mostrar os dados" com o próprio objeto.
    def mostrar_resultados(self):
        # Usamos os métodos da própria classe e formatamos com 2 casas decimais (.2f)
        print(f"Area do terreno = {self.calcular_area():.2f}")
        print(f"Preco do terreno = {self.calcular_preco():.2f}\n")


# ==========================================
# FLUXO PRINCIPAL DO PROGRAMA (FORA DA CLASSE)
# ==========================================

# EXPLICAÇÃO: Entradas de dados padrão, mantendo a familiaridade do estilo procedural.
largura_digitada = float(input("Digite a largura do terreno: "))
comprimento_digitado = float(input("Digite o comprimento do terreno: "))
valor_digitado = float(input("Digite o valor do metro quadrado: "))

# EXPLICAÇÃO: Criação do Objeto (Instanciação).
# INTENÇÃO: Pegamos os valores soltos digitados pelo usuário e os "empacotamos" em um único objeto inteligente chamado 'meu_terreno'.
meu_terreno = Terreno(largura_digitada, comprimento_digitado, valor_digitado)

# EXPLICAÇÃO: O objeto executa o método para exibir suas próprias informações.
# DIFERENÇA PROCEDURAL: Não usamos mais a função print() com variáveis e cálculos soltos. Nós damos a ordem ao objeto, e ele faz todo o trabalho.
meu_terreno.mostrar_resultados()
```
