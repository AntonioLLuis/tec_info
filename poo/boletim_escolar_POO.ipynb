```python
class Aluno:
    """Classe que representa um aluno e encapsula as suas notas e regras de aprovação."""

    def __init__(self, nome, frequencia):
        self.nome = nome
        self.frequencia = frequencia
        self.boletim = {}  # Dicionário para armazenar matéria: lista_de_notas

    def adicionar_notas(self, materia, notas):
        """Adiciona as notas de uma matéria ao boletim do aluno."""
        self.boletim[materia] = notas

    def calcular_media(self, materia):
        """Calcula a média aritmética de uma matéria específica."""
        notas = self.boletim.get(materia, [])
        if not notas:
            return 0.0
        return sum(notas) / len(notas)

    def determinar_situacao(self, materia):
        """Define a situação do aluno na matéria com base na média e frequência geral."""
        media = self.calcular_media(materia)
        if self.frequencia < 75:
            return "Reprovado por Falta"
        elif media >= 7.0:
            return "Aprovado"
        elif media >= 5.0:
            return "Recuperação"
        else:
            return "Reprovado por Nota"

    def exibir_boletim(self, escola="Colégio Avançar de Ensino"):
        """Gera e exibe o relatório visual do boletim na tela."""
        print("\n" + "=" * 50)
        print(f"{escola.upper()}".center(50))
        print("BOLETIM ESCOLAR FINAL".center(50))
        print("=" * 50)
        print(f"Aluno(a): {self.nome}")
        print(f"Frequência Geral: {self.frequencia:.1f}%")
        print("=" * 50)

        for materia, notas in self.boletim.items():
            media = self.calcular_media(materia)
            situacao = self.determinar_situacao(materia)

            notas_formatadas = ", ".join([f"{n:.1f}" for n in notas])
            print(f"Matéria: {materia}")
            print(f"  Notas: [{notas_formatadas}]")
            print(f"  Média: {media:.1f} | Situação: {situacao}")
            print("-" * 50)


class SistemaCadastro:
    """Classe responsável por gerir a interação com o utilizador e o fluxo de entradas."""

    @staticmethod
    def ler_frequencia():
        """Garante que a frequência digitada seja um valor válido entre 0 e 100."""
        while True:
            try:
                freq = float(input("Digite a frequência geral do aluno (%): "))
                if 0 <= freq <= 100:
                    return freq
                print("Por favor, digite um valor entre 0 e 100.")
            except ValueError:
                print("Entrada inválida. Digite apenas números.")

    @staticmethod
    def ler_notas_materia(nome_materia, quantidade_notas=3):
        """Solicita as notas de uma matéria específica ao usuário."""
        notas = []
        print(f"\n--- Notas de {nome_materia} ---")
        for i in range(quantidade_notas):
            while True:
                try:
                    nota = float(input(f"Digite a nota {i+1} (0 a 10): "))
                    if 0 <= nota <= 10:
                        notas.append(nota)
                        break
                    print("A nota deve ser entre 0 e 10.")
                except ValueError:
                    print("Entrada inválida. Digite apenas números.")
        return notas

    @classmethod
    def iniciar_cadastro(cls):
        """Método de fábrica para criar um objeto Aluno preenchido via input."""
        print("=" * 50)
        print("CADASTRO DE ALUNO".center(50))
        print("=" * 50)

        nome = input("Digite o nome do aluno: ").strip()
        frequencia = cls.ler_frequencia()

        # Instancia o objeto Aluno
        aluno = Aluno(nome, frequencia)

        materias = ["Matemática", "História", "Física"]

        # Altera o estado do objeto invocando o seu próprio método
        for materia in materias:
            notas = cls.ler_notas_materia(materia)
            aluno.adicionar_notas(materia, notas)

        return aluno


# --- EXECUÇÃO DO PROGRAMA ---
if __name__ == "__main__":
    # 1. Coleta os dados dinamicamente e cria a instância do objeto
    aluno_cadastrado = SistemaCadastro.iniciar_cadastro()

    # 2. Invocação do comportamento interno do objeto
    aluno_cadastrado.exibir_boletim()676567854327553
```

```python
# A palavra reservada 'class' define um novo tipo de dado (um molde).
# Diferença Procedural: No procedural, não criamos moldes. Os dados (nome, notas)
# ficavam soltos num dicionário e as funções existiam independentemente deles.
class Aluno:

    # O __init__ é o método construtor. Ele é chamado automaticamente ao criar um novo Aluno.
    # O 'self' é a referência ao próprio objeto que está a ser criado.
    # Diferença Procedural: Funções procedurais não têm 'self' nem inicializadores de estado.
    # Elas apenas recebem parâmetros temporários.
    def __init__(self, nome, frequencia):
        # 'self.nome' guarda a variável 'nome' dentro do próprio objeto.
        self.nome = nome

        # 'self.frequencia' guarda a frequência internamente.
        self.frequencia = frequencia

        # 'self.boletim' é inicializado como um dicionário vazio que pertence a este aluno específico.
        # Diferença Procedural: No procedural, este dicionário era criado solto na função main()
        # e passado de mão em mão para outras funções. Aqui, ele "vive" dentro do aluno.
        self.boletim = {}

    # Define um método (comportamento) que pertence ao objeto Aluno.
    def adicionar_notas(self, materia, notas):
        # Acede ao dicionário interno (self.boletim) e adiciona a chave da matéria com a lista de notas.
        # Diferença Procedural: No procedural, alteraríamos o dicionário diretamente no fluxo
        # principal ou numa função genérica recebendo o dicionário como parâmetro.
        self.boletim[materia] = notas

    # Define o método para calcular a média, precisando apenas saber a matéria.
    # Diferença Procedural: A função procedural recebia a 'lista de notas' como argumento.
    # Aqui, o objeto já tem as notas guardadas em 'self.boletim', então só precisa do nome da matéria.
    def calcular_media(self, materia):
        # Vai buscar as notas internamente usando self.boletim.get(). Retorna lista vazia [] se não existir.
        notas = self.boletim.get(materia, [])

        # Verifica se a lista está vazia. Se sim, retorna 0.0.
        if not notas:
            return 0.0

        # Calcula e retorna a média aritmética.
        return sum(notas) / len(notas)

    # Define o método que determina a situação, acedendo às regras de negócio.
    def determinar_situacao(self, materia):
        # O objeto invoca outro método próprio (self.calcular_media) para obter a nota.
        # Diferença Procedural: No procedural, a média e a frequência tinham de ser
        # passadas como argumentos: determinar_situacao(media, frequencia).
        # Aqui, o objeto "lê" a sua própria frequência (self.frequencia) e calcula a sua média.
        media = self.calcular_media(materia)

        # Verifica as condições e retorna a string de estado correspondente.
        if self.frequencia < 75:
            return "Reprovado por Falta"
        elif media >= 7.0:
            return "Aprovado"
        elif media >= 5.0:
            return "Recuperação"
        else:
            return "Reprovado por Nota"

    # Define o método de apresentação visual. Pode receber parâmetros opcionais.
    def exibir_boletim(self, escola="Colégio Avançar de Ensino"):
        # Imprime o cabeçalho formatado.
        print("\n" + "=" * 50)
        print(f"{escola.upper()}".center(50))
        print("BOLETIM ESCOLAR FINAL".center(50))
        print("=" * 50)

        # Acede aos próprios dados internos (self.nome e self.frequencia) para imprimir.
        # Diferença Procedural: No procedural, fazíamos print(aluno['nome']), lendo de um dicionário solto.
        print(f"Aluno(a): {self.nome}")
        print(f"Frequência Geral: {self.frequencia:.1f}%")
        print("=" * 50)

        # Percorre o próprio dicionário interno (self.boletim).
        for materia, notas in self.boletim.items():
            # O objeto calcula a própria média e determina a própria situação usando os seus métodos.
            media = self.calcular_media(materia)
            situacao = self.determinar_situacao(materia)

            # Formata as notas para exibição visual.
            notas_formatadas = ", ".join([f"{n:.1f}" for n in notas])

            # Imprime os resultados calculados internamente.
            print(f"Matéria: {materia}")
            print(f"  Notas: [{notas_formatadas}]")
            print(f"  Média: {media:.1f} | Situação: {situacao}")
            print("-" * 50)


# Nova classe criada apenas para organizar as funções que lidam com input do utilizador.
# Diferença Procedural: No procedural, estas funções ficavam espalhadas soltas no arquivo.
# Aqui, estão agrupadas de forma lógica dentro do contexto do 'SistemaCadastro'.
class SistemaCadastro:

    # O decorador @staticmethod indica que este método não precisa de ler nem modificar
    # o estado da classe ou do objeto. Ele funciona como uma função procedural clássica,
    # mas está guardado aqui dentro por organização.
    @staticmethod
    def ler_frequencia():
        # Entra num loop infinito até o utilizador digitar um valor válido.
        while True:
            try:
                # Solicita o input e tenta converter para número decimal (float).
                freq = float(input("Digite a frequência geral do aluno (%): "))

                    return freq
                # Caso não seja, dá feedback de erro.
                print("Por favor, digite um valor entre 0 e 100.")
            except ValueError:
                # Captura o erro caso o utilizador digite letras em vez de números.
                print("Entrada inválida. Digite apenas números.")

    # Outro método estático, pois não acede ao estado da classe (não tem 'self').
    @staticmethod
    def ler_notas_materia(nome_materia, quantidade_notas=3):
        # Inicializa a lista de notas vazia localmente.
        notas = []
        print(f"\n--- Notas de {nome_materia} ---")

        # Repete para a quantidade de notas exigida.
        for i in range(quantidade_notas):
            while True:
                try:
                    # Coleta cada nota e tenta converter para float.
                    nota = float(input(f"Digite a nota {i+1} (0 a 10): "))
                    # Valida se está entre 0 e 10.
                    if 0 <= nota <= 10:
                        # Adiciona a lista e quebra o loop interior (break).
                        notas.append(nota)
                        break
                    print("A nota deve ser entre 0 e 10.")
                except ValueError:
                    print("Entrada inválida. Digite apenas números.")
        # Retorna a lista preenchida.
        return notas

    # O decorador @classmethod passa a própria classe (cls) como primeiro argumento,
    # e não a instância. Usamos isto muitas vezes para "fabricar" objetos.
    @classmethod
    def iniciar_cadastro(cls):
        # Exibe cabeçalho.
        print("=" * 50)
        print("CADASTRO DE ALUNO".center(50))
        print("=" * 50)

        # Recolhe o nome.
        nome = input("Digite o nome do aluno: ").strip()
        # Invoca o método estático desta mesma classe (cls) para ler a frequência.
        frequencia = cls.ler_frequencia()

        # AQUÍ É A GRANDE DIFERENÇA:
        # Em vez de criar um dicionário comum como no procedural,
        # Instanciamos (criamos) um objeto do tipo 'Aluno' chamado 'aluno'.
        aluno = Aluno(nome, frequencia)

        # Lista de matérias para o loop.
        materias = ["Matemática", "História", "Física"]

        for materia in materias:
            # Recolhe as notas invocando o método da classe.
            notas = cls.ler_notas_materia(materia)
            # Envia as notas para DENTRO do objeto 'aluno', usando o método dele.
            aluno.adicionar_notas(materia, notas)

        # Retorna o objeto completo, vivo, carregado de dados e métodos.
        # Diferença Procedural: O procedural retornava apenas um dicionário "burro" com os dados.
        return aluno


# Esta condição verifica se o ficheiro está a ser executado diretamente (não importado).
if __name__ == "__main__":
    # 1. Chamamos o sistema de cadastro. Ele interage com o utilizador, cria o
    #    objeto Aluno internamente, e devolve-o para a variável 'aluno_cadastrado'.
    aluno_cadastrado = SistemaCadastro.iniciar_cadastro()

    # 2. Invocação do comportamento.
    # Diferença Procedural: No procedural, fazíamos exibir_boletim(aluno_cadastrado).
    # O foco era a FUNÇÃO, que recebia o dado passivamente.
    # Na POO, o foco é o OBJETO: pedimos ao próprio 'aluno_cadastrado' que exiba
    # o seu boletim: objeto.metodo()
    aluno_cadastrado.exibir_boletim()
```
