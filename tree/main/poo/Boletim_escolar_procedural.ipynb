```python
# --- FUNÇÕES DE PROCESSAMENTO (Lógica) ---

def calcular_media(notas):
    """Calcula a média aritmética de uma lista de notas."""
    if not notas:
        return 0.0
    return sum(notas) / len(notas)


def determinar_situacao(media, frequencia):
    """Define se o aluno foi aprovado, ficou de recuperação ou reprovado."""
    if frequencia < 75:
        return "Reprovado por Falta"
    elif media >= 7.0:
        return "Aprovado"
    elif media >= 5.0:
        return "Recuperação"
    else:
        return "Reprovado por Nota"


# --- FUNÇÕES DE ENTRADA DE DADOS (Inputs) ---

def ler_frequencia():
    """Garante que a frequência digitada seja um valor válido entre 0 e 100."""
    while True:
        try:
            freq = float(input("Digite a frequência geral do aluno (%): "))
            if 0 <= freq <= 100:
                return freq
            print("Por favor, digite um valor entre 0 e 100.")
        except ValueError:
            print("Entrada inválida. Digite apenas números.")


def ler_notas_materia(nome_materia, quantidade_notas=3):
    """Solicita as notas de uma matéria específica ao usuário."""
    notas = []
    print(f"\n--- Notas de {nome_materia} ---")
    for i in range(quantidade_notas):
        while True:
            try:
                nota = float(input(f"Digite a nota {i+1} (0 a 10): "))
                if 0 <= nota <= 10:
                    notas.append(nota)
                    break
                print("A nota deve ser entre 0 e 10.")
            except ValueError:
                print("Entrada inválida. Digite apenas números.")
    return notas


def cadastrar_aluno_e_notas():
    """Procedimento para coletar todos os dados do aluno via input."""
    print("=" * 50)
    print("CADASTRO DE ALUNO".center(50))
    print("=" * 50)

    nome = input("Digite o nome do aluno: ").strip()
    frequencia = ler_frequencia()

    # Lista de matérias fixa para o exemplo (pode ser alterada)
    materias = ["Matemática", "História", "Física"]
    notas_por_materia = {}

    for materia in materias:
        notas_por_materia[materia] = ler_notas_materia(materia)

    # Retorna a estrutura de dados preenchida pelo usuário
    return {
        "nome": nome,
        "frequencia": frequencia,
        "notas_por_materia": notas_por_materia
    }


# --- FUNÇÕES DE APRESENTAÇÃO (Visual) ---

def exibir_boletim(aluno, escola="Colégio Avançar de Ensino"):
    """Gera o relatório visual do boletim na tela."""
    print("\n" + "=" * 50)
    print(f"{escola.upper()}".center(50))
    print("BOLETIM ESCOLAR FINAL".center(50))
    print("=" * 50)
    print(f"Aluno(a): {aluno['nome']}")
    print(f"Frequência Geral: {aluno['frequencia']:.1f}%")
    print("=" * 50)

    for materia, notas in aluno["notas_por_materia"].items():
        media = calcular_media(notas)
        situacao = determinar_situacao(media, aluno["frequencia"])

        notas_formatadas = ", ".join([f"{n:.1f}" for n in notas])
        print(f"Matéria: {materia}")
        print(f"  Notas: [{notas_formatadas}]")
        print(f"  Média: {media:.1f} | Situação: {situacao}")
        print("-" * 50)


# --- EXECUÇÃO DO PROGRAMA ---

if __name__ == "__main__":
    # 1. Coleta os dados dinamicamente com inputs
    dados_aluno = cadastrar_aluno_e_notas()

    # 2. Exibe o boletim processado
    exibir_boletim(dados_aluno)
```

```python
"""
================================================================================
PARADIGMA USADO: PROGRAMAÇÃO PROCEDURAL (vs. POO)
================================================================================
Este script foi construído estritamente sob o paradigma PROCEDURAL.
Diferença prática aplicada neste código:
1. Separação de Dados e Lógica: Os dados do aluno são estruturas 'muda'
   (dicionários/listas). As funções existem isoladamente fora dos dados.
2. Passagem Explícita: As funções recebem os dados por parâmetros (ex: 'aluno').
   Em POO, o próprio objeto invocaria o método e usaria seu estado interno (self).
3. Decomposição Top-Down: O problema foi dividido em rotinas e sub-rotinas.
4. Sem Classes: Não existem moldes de classes (class Aluno) ou objetos instanciados.
================================================================================
"""

# --- FUNÇÕES DE PROCESSAMENTO (Lógica) ---

# Declara a função que aceita a lista de notas por parâmetro
def calcular_media(notas):
    """Calcula a média aritmética de uma lista de notas."""
    # Se a lista estiver vazia (not notas), retorna 0.0 prevenindo erro de divisão por zero
    if not notas:
        return 0.0
    # sum(notas) soma os elementos e len(notas) conta a quantidade para obter a média
    return sum(notas) / len(notas)


# Declara a função que avalia a aprovação recebendo a média e a frequência
def determinar_situacao(media, frequencia):
    """Define se o aluno foi aprovado, ficou de recuperação ou reprovado."""
    # Regra 1: Frequência abaixo de 75% reprova direto por falta
    if frequencia < 75:
        return "Reprovado por Falta"
    # Regra 2: Frequência OK e média >= 7.0 aprova
    elif media >= 7.0:
        return "Aprovado"
    # Regra 3: Frequência OK e média entre 5.0 e 6.9 vai para recuperação
    elif media >= 5.0:
        return "Recuperação"
    # Regra 4: Frequência OK e média menor que 5.0 reprova por nota
    else:
        return "Reprovado por Nota"


# --- FUNÇÕES DE ENTRADA DE DADOS (Inputs) ---

# Função sem parâmetros para leitura da frequência
def ler_frequencia():
    """Garante que a frequência digitada seja um valor válido entre 0 e 100."""
    # Loop infinito que só é interrompido ao obter um 'return' válido
    while True:
        try:
            # Tenta converter o texto digitado pelo usuário em número decimal (float)
            freq = float(input("Digite a frequência geral do aluno (%): "))
            # Verifica se o número está no intervalo permitido [0, 100]
            if 0 <= freq <= 100:
                return freq  # Retorna o valor e encerra a função
            # Exibe aviso caso o número esteja fora da faixa esperada
            print("Por favor, digite um valor entre 0 e 100.")
        except ValueError:
            # Captura o erro caso o usuário digite letras ou símbolos não numéricos
            print("Entrada inválida. Digite apenas números.")


# Função para ler notas de uma matéria específica (com parâmetro opcional quantidade_notas=3)
def ler_notas_materia(nome_materia, quantidade_notas=3):
    """Solicita as notas de uma matéria específica ao usuário."""
    # Inicializa uma lista vazia para armazenar as notas da matéria
    notas = []
    # Imprime o cabeçalho da matéria
    print(f"\n--- Notas de {nome_materia} ---")
    # Repete a leitura a quantidade de vezes definida (por padrão, 3)
    for i in range(quantidade_notas):
        # Loop de validação individual para a nota atual
        while True:
            try:
                # Solicita a nota exibindo o índice amigável i+1 (ex: nota 1, nota 2...)
                nota = float(input(f"Digite a nota {i+1} (0 a 10): "))
                # Valida se a nota está entre 0 e 10
                if 0 <= nota <= 10:
                    notas.append(nota)  # Insere a nota válida na lista 'notas'
                    break               # Encerra o 'while True' e avança o loop 'for'
                print("A nota deve ser entre 0 e 10.")
            except ValueError:
                # Trata erro de entrada não numérica
                print("Entrada inválida. Digite apenas números.")
    # Retorna a lista contendo as notas coletadas
    return notas


# Função procedural central de cadastro de dados
def cadastrar_aluno_e_notas():
    """Procedimento para coletar todos os dados do aluno via input."""
    # Exibe cabeçalho de cadastro formatado na tela
    print("=" * 50)
    print("CADASTRO DE ALUNO".center(50))
    print("=" * 50)

    # Lê o nome do aluno e remove espaços nas pontas (.strip())
    nome = input("Digite o nome do aluno: ").strip()
    # Invoca a função 'ler_frequencia' e guarda o retorno na variável
    frequencia = ler_frequencia()

    # Lista com as matérias padrão a serem processadas
    materias = ["Matemática", "História", "Física"]
    # Dicionário vazio para associar cada matéria à sua lista de notas
    notas_por_materia = {}

    # Percorre cada matéria contida na lista
    for materia in materias:
        # Popula o dicionário chamando a função de leitura para a matéria atual
        notas_por_materia[materia] = ler_notas_materia(materia)

    # Retorna a estrutura de dados no formato de dicionário
    return {
        "nome": nome,
        "frequencia": frequencia,
        "notas_por_materia": notas_por_materia
    }


# --- FUNÇÕES DE APRESENTAÇÃO (Visual) ---

# Função responsável por renderizar o boletim na tela
def exibir_boletim(aluno, escola="Colégio Avançar de Ensino"):
    """Gera o relatório visual do boletim na tela."""
    # Imprime o cabeçalho do relatório formatado
    print("\n" + "=" * 50)
    print(f"{escola.upper()}".center(50))  # Transforma o nome em maiúsculas e centraliza
    print("BOLETIM ESCOLAR FINAL".center(50))
    print("=" * 50)
    # Acessa os dados diretamente das chaves do dicionário 'aluno'
    print(f"Aluno(a): {aluno['nome']}")
    print(f"Frequência Geral: {aluno['frequencia']:.1f}%")  # Formata float para 1 casa decimal
    print("=" * 50)

    # Itera sobre os pares chave (materia) e valor (notas) de 'notas_por_materia'
    for materia, notas in aluno["notas_por_materia"].items():
        # Chama a função lógica 'calcular_media' passando as notas da matéria corrente
        media = calcular_media(notas)
        # Chama 'determinar_situacao' enviando a média calculada e a frequência do aluno
        situacao = determinar_situacao(media, aluno["frequencia"])

        # Converte a lista de floats em texto formatado separado por vírgulas
        notas_formatadas = ", ".join([f"{n:.1f}" for n in notas])
        # Imprime as informações consolidadas da matéria
        print(f"Matéria: {materia}")
        print(f"  Notas: [{notas_formatadas}]")
        print(f"  Média: {media:.1f} | Situação: {situacao}")
        print("-" * 50)


# --- EXECUÇÃO DO PROGRAMA ---

# Bloco condicional que verifica se o script está sendo executado diretamente
if __name__ == "__main__":
    # 1. Executa a rotina de cadastro e recebe a estrutura de dados montada
    dados_aluno = cadastrar_aluno_e_notas()

    # 2. Envia a estrutura de dados para a rotina de exibição do boletim
    exibir_boletim(dados_aluno)
```
